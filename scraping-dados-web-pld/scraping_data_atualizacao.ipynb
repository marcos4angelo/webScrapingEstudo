```python
from urllib.request import urlopen
from urllib.error import HTTPError
from urllib.error import URLError
from bs4 import BeautifulSoup
import pandas as pd
import re
```

```python
def get_bs_html(url):
    """
    This function generates a BeautifulSoup object to scrape through.
    """
    try:
        html = urlopen(url)
    except HTTPError as e:
        print(e)
        return None
    except URLError as e:
        print('The server could not be found.')
    try:
        bs = BeautifulSoup(html.read(), 'html.parser')
    except:
        return None
    return bs

def generate_table_rfb(soup):
    """
    Extracts folder names (YYYY-MM) and their last updated timestamps from a BeautifulSoup object.
    """
    
    rows = soup.find_all("tr")
    pattern = re.compile(r"^\d{4}-\d{2}$")  # Match YYYY-MM format
    data = []

    for row in rows:
        cells = row.find_all("td")
        if len(cells) >= 3:  # Ensure there are at least 3 <td> elements
            folder_name = cells[1].find("a").text.strip("/")  # Extract folder name
            timestamp = cells[2].text.strip()  # Extract timestamp

            if pattern.match(folder_name):  # Keep only YYYY-MM formatted names
                data.append((folder_name, timestamp))

    return pd.DataFrame(data, columns=["companies_folder_date", "updated_at"])
```

```python
bs = get_bs_html('https://arquivos.receitafederal.gov.br/dados/cnpj/dados_abertos_cnpj/')
```

```python
df = generate_table_rfb(bs)
```
